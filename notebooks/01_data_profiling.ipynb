```python
import pandas as pd
import numpy as np

# Load raw dataset
raw_path = '../data/raw/Supermart Grocery Sales - Retail Analytics Dataset.csv'
df = pd.read_csv(raw_path)

print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
# Check for nulls and duplicate rows
print("Null Values Count:\n", df.isnull().sum())
print(f"\nDuplicate Rows: {df.duplicated().sum()}")
print("\nUnique Values Count:\n", df.nunique())
```

```python
# 1. Clean column names to standard snake_case
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

# 2. Standardize mixed date strings (slashes and hyphens) into datetime
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed', dayfirst=False)

# 3. Round numerical fields for precision consistency
df['discount'] = df['discount'].round(2)
df['profit'] = df['profit'].round(2)

# Verify cleaned schema
df.info()
```

```python
# Export cleaned dataset for SQL Server ingestion
clean_path = '../data/processed/cleaned_supermart_sales.csv'
df.to_csv(clean_path, index=False)
print(f"Cleaned dataset saved successfully to {clean_path}")
```
